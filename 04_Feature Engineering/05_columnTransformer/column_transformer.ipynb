## Column Transformer

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.model_selection import train_test_split
```

```python
X =df.drop(columns='has_covid',axis=1)
y = df['has_covid']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
col_transformer = ColumnTransformer(transformers=[
    ('simple_imputer', SimpleImputer(), ['fever']),
    ('ordinal_encoder', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),
    ('one_hot_encoder', OneHotEncoder(sparse_output=False, drop='first'), ['gender', 'city'])
], remainder='passthrough')
```

```python
col_transformer.fit_transform(X_train)
```

```python
col_transformer.transform(X_test)
```
